```python
from abc import ABC, abstractmethod

class LibraryItem(ABC):
  def __init__(self,item_id,title):
    self.item_id = item_id
    self.title = title

  @abstractmethod
  def calculate_late_fee(self,days):
    pass

class Book(LibraryItem):
  def calculate_late_fee(self,days):
    return days * 10

class Magazine(LibraryItem):
  def calculate_late_fee(self,days):
    return days * 5

def late_fee(item,days):
  if days < 0:
    print("Invalid number of days!")

  fee = item.calculate_late_fee(days)

  print("\n----- Library Details -----")
  print("Item ID:", item.item_id)
  print("Title:", item.title)
  print("Late Days:", days)
  print("Late Fee: Rs.", fee)

try:
  print("1. Book")
  print("2. Magazine")
  choice = int(input("Enter your choice: "))

  item_id = input("Enter Item ID: ")
  title = input("Enter Item Title: ")
  days = int(input("Enter Late Days: "))

  item = None
  if choice == 1:
    item = Book(item_id,title)

  elif choice == 2:
    item = Magazine(item_id,title)

  else:
    print("Invalid choice!")

  if item is not None:
    late_fee(item, days)

except ValueError:
    print("Please enter valid input!")
```
